```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("employees_cleaned2_5000.csv")

fig, ax = plt.subplots(2, 3, figsize=(18, 10))

plt.suptitle("Employee Analytics Dashboard", fontsize=18)

rating = df["PerformanceRating"].value_counts()
ax[0,0].pie(rating.values, labels=rating.index, autopct="%1.1f%%")
ax[0,0].set_title("Performance Rating")

gender = df["Gender"].value_counts()
ax[0,1].pie(gender.values, labels=gender.index, autopct="%1.1f%%")
ax[0,1].set_title("Gender")

dept = df.groupby("Department")["Salary"].mean()
ax[0,2].barh(dept.index, dept.values, color="purple", edgecolor="black")
ax[0,2].set_title("Average Salary")
ax[0,2].set_xlabel("Salary")

ax[1,0].hist(df["Salary"], bins=20, color="skyblue", edgecolor="black")
ax[1,0].set_title("Salary Distribution")
ax[1,0].set_xlabel("Salary")
ax[1,0].set_ylabel("Employees")

exp = df.groupby("YearsOfExperience")["Salary"].mean()
ax[1,1].plot(exp.index, exp.values, marker="o", color="red")
ax[1,1].set_title("Salary vs Experience")
ax[1,1].set_xlabel("Years of Experience")
ax[1,1].set_ylabel("Average Salary")
ax[1,1].grid(True)

city = df["City"].value_counts().head(10)
ax[1,2].barh(city.index, city.values, color="orange", edgecolor="black")
ax[1,2].set_title("Top 10 Cities")
ax[1,2].set_xlabel("Employees")

plt.savefig("Employee Analytics Dashboard")
plt.tight_layout()
plt.show()
```
